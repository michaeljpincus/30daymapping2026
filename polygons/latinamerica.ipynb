```python
import geopandas as gpd
countries = gpd.read_file('countries_lakes/ne_10m_admin_0_countries_lakes.shp')
```

```python
map_countries = countries[countries['NAME'].isin(['Brazil', 'Chile', 'Colombia', 'Peru', 'Costa Rica', 'Honduras'])]
```

```python
map_countries.to_file('map_countries.geojson', driver='GeoJSON')
```
